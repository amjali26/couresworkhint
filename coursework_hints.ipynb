```python
import pandas as pd

# TODO: Define the base URL to the GitHub repository where the CSV files are stored
base_url = "https://raw.githubusercontent.com/XiangX91/CEG3005_25/refs/heads/main/Coursework/"

# TODO: Choose the year you're interested in
year = 9  # <-- Fill in the year as a single digit or two-digit number

# TODO: Construct the year string in the format '20XX-YY'
yr_str = '20' + str(year).zfill(2) + '-' + str(year + 1).zfill(2)

# Optional: Print to check your formatted year string
print("Year string:", yr_str)

# TODO: Build the full URL to the dataset file
file_url = base_url + f"estimates-of-station-usage-{yr_str}.csv"

# TODO: Load the dataset from the URL into a pandas DataFrame
df = pd.read_csv(file_url)

# Preview the first few rows of the DataFrame
print(df.head())
```

### 🔗 Merging Two Years of Station Usage Data

This example shows how to **combine data from two different financial years** using the common column `"Station Name"`.

#### 🧾 What the code does:
- Loads **2009–10** and **2010–11** datasets from a GitHub repository using `pandas.read_csv()`.
- Prints the column names of both DataFrames to help you verify they can be merged (they must share a common column).
- Merges the two DataFrames using `pd.merge()` on `"Station Name"`.
- Shows the first few rows of the merged result.

---

#### ❗🔴 **IMPORTANT NOTE – PLEASE READ**
> This code only shows how to merge **two years** of data.  
> Since you need to process **14 years**, this is **not the best structure**.  
> ❌ **Do NOT just copy and paste this for all years.**  
> ✅ Instead, you should write a **loop** or a **function** to load and merge all 14 datasets automatically.

---

This example is just to help you understand the merging logic — you’ll improve and scale it for your final solution.

```python
import pandas as pd

# Define the URL for the 2009–10 dataset
url1 = 'https://raw.githubusercontent.com/XiangX91/CEG3005_25/refs/heads/main/Coursework/estimates-of-station-usage-2009-10.csv'

# Define the URL for the 2010–11 dataset
url2 = 'https://raw.githubusercontent.com/XiangX91/CEG3005_25/refs/heads/main/Coursework/estimates-of-station-usage-2010-11.csv'

# Load each dataset into a separate DataFrame
df1 = pd.read_csv(url1)
df2 = pd.read_csv(url2)

# Print column names of both datasets to check if they have a common key for merging
print(df1.columns)
print(df2.columns)

# Merge the two datasets on the "Station Name" column (i.e. keep only rows with matching station names)
merged_df = pd.merge(df1, df2, on='Station Name')

# Display the first few rows of the merged dataset
merged_df.head()
```
